```python
import pandas as pd

df_customers = pd.read_csv('data/customers.csv')
df_products = pd.read_csv('data/articles.csv')
df_transactions = pd.read_csv('data/transactions_train.csv')

df_customers.to_parquet('data/customers.parquet', index=False)
df_products.to_parquet('data/articles.parquet', index=False)
df_transactions.to_parquet('data/transactions_train.parquet', index=False)
```

```python
from pathlib import Path
import pandas as pd

# Path.cwd() te da la ruta actual de Jupyter (seguramente acabe en .../analysis/utils)
# Con .parent subes a 'analysis', y con el segundo .parent subes a la raíz del proyecto
BASE_DIR = Path.cwd().parent.parent

# Ahora entramos en la carpeta data
DATA_DIR = BASE_DIR / 'data'

# Tus archivos
archivos_csv = ['articles.csv', 'customers.csv', 'transactions_train.csv']

print(f"Directorio base detectado: {BASE_DIR}")
print(f"Carpeta de datos detectada: {DATA_DIR}\n")
for archivo in archivos_csv:
    # 1. Al unir DATA_DIR (que es un Path) con archivo, ruta_csv se convierte en un objeto Path
    ruta_csv = DATA_DIR / archivo
    
    # 2. Ahora sí podemos usar .with_suffix() sobre ruta_csv
    ruta_parquet = ruta_csv.with_suffix('.parquet')
    
    if ruta_csv.exists():
        print(f"⏳ Leyendo: {archivo}...")
        df = pd.read_csv(ruta_csv)
        df.to_parquet(ruta_parquet, index=False, engine='pyarrow')
        print(f"✅ ¡Éxito! Guardado como: {ruta_parquet.name}\n")
    else:
        print(f"❌ Error: No se encontró el archivo en {ruta_csv}\n")
```
